```python
import cv2
import matplotlib.pyplot as plt


img = cv2.imread("image.jpg")
# Chuyển sang ảnh xám (cần cho cascade)
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
# Chuyển sang RGB để hiển thị đúng màu với matplotlib
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Tải cascade đã được huấn luyện cho biển báo STOP (file stop_data.xml)
stop_data = cv2.CascadeClassifier("stop_data.xml")

# Phát hiện biển báo với kích thước tối thiểu 20x20 pixel
found = stop_data.detectMultiScale(img_gray, minSize=(20, 20))
amount_found = len(found)
if amount_found != 0:

    for x, y, width, height in found:
        cv2.rectangle(img_rgb, (x, y), (x + height, y + width), (0, 255, 0), 5)
plt.imshow(img_rgb)
plt.show() 
```

```python
img = cv2.imread("image_2.jpg")
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Tải cascade cho khuôn mặt (file XML đã tải về)
haarcascade = cv2.CascadeClassifier("haarcascade_frontalface_default.xml")

# Phát hiện khuôn mặt
found = haarcascade.detectMultiScale(img_gray, minSize=(20, 20))
amount_found = len(found)

if amount_found != 0:
    for x, y, width, height in found:
        cv2.rectangle(img_rgb, (x, y), (x + height, y + width), (0, 255, 0), 5)

plt.imshow(img_rgb)
plt.show()
```
